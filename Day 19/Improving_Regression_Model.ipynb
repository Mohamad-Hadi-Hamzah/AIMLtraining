```python
import pandas as pd
import numpy as np
from math import sqrt
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
```

```python
data = {
    'StudyHours': [2, 3, 4, 5, 6, 7, 8, 9, 10],
    'Attendance': [60, 65, 70, 72, 75, 80, 85, 88, 90],
    'PrevScore': [40, 45, 50, 55, 60, 65, 70, 75, 80],
    'FinalMarks': [50, 55, 60, 65, 70, 75, 78, 85, 88]
}
df = pd.DataFrame(data)
df
```

```python
y=df['FinalMarks'] #output
y
```

    # 🔹 Purpose
    
    # This is called a baseline model.
    # It’s the simplest possible prediction: always predict the average.
    # You can then compare your actual regression model to this baseline to see if it’s actually learning something useful.
    
    # Plain English:
    # “Predict the average value for every row. This is the simplest benchmark to measure model performance.”

```python
y_pred_baseline = np.repeat(y.mean(), len(y))
```

```python
# Calculate metrics
mae = mean_absolute_error(y, y_pred_baseline)
mse = mean_squared_error(y, y_pred_baseline)
rmse = np.sqrt(mse)
r2 = r2_score(y, y_pred_baseline)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R²:", r2)
```
