# Predict Heart Disease
  1.Load Dataset
  2.Split Features and Test
  3.Train Test Split
  4.Scale Features

```python
import pandas as pd
import numpy as np
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
```

```python
heart_data=pd.read_csv('/content/sample_data/heart.csv')
heart_data.head()
```

```python
x=heart_data.drop('target',axis=1)
y=heart_data['target']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
scalar=StandardScaler()
x_train_scaled=scalar.fit_transform(x_train)
x_test_scaled=scalar.transform(x_test)
```

```python
model=tf.keras.Sequential(
    [
     tf.keras.layers.Dense(16,activation='relu',input_shape=(x_train.shape[1],)),
     tf.keras.layers.Dense(8,activation='relu'),
     tf.keras.layers.Dense(1,activation='sigmoid')
])
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
#train the model
history_model=model.fit(x_train,y_train,epochs=100,validation_data=(x_test,y_test))
```

```python
# Evaluate the model
loss,accuracy=model.evaluate(x_test,y_test)
print(f'Test Loss: {loss:.4f}')
print(f'Test Accuracy: {accuracy*100:.2f}%')
```

# Example 2
#*Bank Customer Exit Prediction*
#*Based on given data using Neural Network Binary Classifier find out if customer staying or leaving:(Churn=0)and leave(Churn=1)

```python
bank_data=pd.read_csv('/content/sample_data/bank_churn.csv')
bank_data.head()
```

```python
le = LabelEncoder()
bank_data['Gender'] = le.fit_transform(bank_data['Gender'])
bank_data=pd.get_dummies(bank_data,columns=['Geography'])
```

```python
x=bank_data.drop('Exited',axis=1)
y=bank_data['Exited']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
model=tf.keras.Sequential(
    [
     tf.keras.layers.Dense(16,activation='relu',input_shape=(x_train.shape[1],)),
     tf.keras.layers.Dense(8,activation='relu'),
     tf.keras.layers.Dense(1,activation='sigmoid')
])
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
history_model=model.fit(x_train,y_train,epochs=100,validation_data=(x_test,y_test))
```

```python
loss,accuracy=model.evaluate(x_test,y_test)
print(f'Test Loss: {loss:.4f}')
print(f'Test Accuracy: {accuracy*100:.2f}%')
```
