```python
import numpy as np
numpy_Array=np.arange(1,25,dtype=np.int16)
numpy_Array
```

```python
tensor_numpy=tf.constant(numpy_Array,shape=(2,3,4))
tensor_numpy
```

```python
print('Tensor Information :')
print('Shape: ',tensor_numpy.shape)
print('Dimension:',tensor_numpy.ndim)
print('Rank: ',tf.rank(tensor_numpy).numpy())
print('Size: ',tf.size(tensor_numpy))
print('DataType of elements: ',tensor_numpy.dtype)
print('Element along axis 0: ',tensor_numpy.shape[0])
print('Element along axis 1: ',tensor_numpy.shape[1])
print('Element along axis 2: ',tensor_numpy.shape[2])
print('Total number of elements: ',tf.size(tensor_numpy).numpy())
```

**Arithmetic Operations with tensors**
  - Add
  - Sub
  - Multi
  - Div

```python
tensor_one=tf.constant([[11,9],[2,5]])
tensor_one
```

```python
#tensor_one+15
tf.add(tensor_one,15)
```

```python
#tensor_one-5
tf.subtract(tensor_one,5)
```

```python
#tensor_one*2
tf.multiply(tensor_one,2)
```

```python
#tensor_one/5
tf.divide(tensor_one,5)
```

```python
tensor_new=tf.square(tensor_one)
tensor_new
```
