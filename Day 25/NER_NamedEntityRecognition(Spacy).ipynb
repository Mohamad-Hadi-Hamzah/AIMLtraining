```python
import spacy
nlp=spacy.load('en_core_web_sm')
print('spacy loaded successfully')
```

```python
#text='Bill Gates founded Microsoft in year 1994 in California - USA'
```

```python
text='Lee Chong WEi is a badminton player from Malaysia. He has won Sudirman Cup in the year 2000'
```

```python
doc=nlp(text)
for ent in doc.ents:
  print(ent.text,ent.label_ ,spacy.explain(ent.label_))
```

```python
from spacy import displacy
displacy.render(doc,style='ent',jupyter=True)
```

```python
corpus=[
    'Taj Mahal is located in Agra India',
    'Google was founded by Larry Page',
    'Twin Towers in located in Kuala Lumpur',
    'Discovery of india was written by Pt. Nehru'
]
```

```python
for text in corpus:
  doc=nlp(text)
  print(f'\n Text: {text}')
  for ent in doc.ents:
    print(ent.text,ent.label_)
```
