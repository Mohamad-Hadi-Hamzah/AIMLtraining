```python
# Import all required libraries

import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

import matplotlib.pyplot as plt
import seaborn as sns
from numpy import vectorize
```

```python
df=pd.read_csv('/content/sample_data/UpdatedResumeDataSet.csv')
df.shape
```

```python
df.head()
```

```python
print('\n Columns',df.columns.tolist())
```

```python
sample_df=df.sample(10,random_state=42).reset_index(drop=True)
sample_df.head()
```

```python
sample_df=sample_df.rename(columns={'Resume':'resume_text'})
sample_df.shape
```

```python
vectorizer=TfidfVectorizer(stop_words='english')
tfidf_matrix=vectorizer.fit_transform(sample_df['resume_text'])
tfidf_matrix.shape
```

```python
similarities_matrix=cosine_similarity(tfidf_matrix)
sim_df=pd.DataFrame(similarities_matrix,index=sample_df['Category']+'_'+sample_df.index.astype(str),
                    columns=sample_df['Category']+'_'+sample_df.index.astype(str))
sim_df.round(2).head(3)
```
