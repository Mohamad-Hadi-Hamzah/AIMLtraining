** NLP PIPELINE **

```python
# Import requires libraries
import nltk
import spacy
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
import re
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')
```

```python
our_text='Artificial Intelligence and Machine Learning are going to change the future of Technology!'
```

```python
tokens=word_tokenize(our_text)
# splits text into individual words or tokens
tokens
```

```python
#import re to clean tokens
clean_tokens=[re.sub('[^a-zA-Z]','',token.lower()) for token in tokens if token.isalpha()]
clean_tokens
```

```python
# stop words removal
stop_words=set(stopwords.words('english'))
filtered_tokens=[token for token in clean_tokens if token not in stop_words]
# removes common less meaningful words
filtered_tokens
```

```python
# stemming
stemmer=PorterStemmer()
stemmed_tokens=[stemmer.stem(token) for token in filtered_tokens]
# reduce the words to their root form (not always a word from dictionary)
stemmed_tokens
```

```python
lemmatizer=WordNetLemmatizer()
lemmatized_tokens=[lemmatizer.lemmatize(token) for token in filtered_tokens]
# reduce the words to their root form (always a word from dictionary)
lemmatized_tokens
```

** Bag of Words (BOW)**

```python
from nltk import corpus
corpus=[
    'Artificial Intelligence is tranforming industries',
    'Machine Learning is a branch of artificial intelligence',
    'Deep Learning is a subfield of machine learning',
    'Natural Language Processing is a subfield of artificial intelligence'
]
```

```python
cv=CountVectorizer()
bow_matrix=cv.fit_transform(corpus)
print('*******Feature names (BOW)*******')
print(cv.get_feature_names_out())
print('*******BOW Matrix Array*******')
print(bow_matrix.toarray())
#
```
